```python
# Define correct data types
dtype = {
    "VendorID": "Int64",
    "passenger_count": "Int64",
    "trip_distance": "float64",
    "RatecodeID": "Int64",
    "store_and_fwd_flag": "string",
    "PULocationID": "Int64",
    "DOLocationID": "Int64",
    "payment_type": "Int64",
    "fare_amount": "float64",
    "extra": "float64",
    "mta_tax": "float64",
    "tip_amount": "float64",
    "tolls_amount": "float64",
    "improvement_surcharge": "float64",
    "total_amount": "float64",
    "congestion_surcharge": "float64"
}

# Columns to parse as dates
parse_dates = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime"
]

print("✅ Data types defined!")
```

```python
# Read with correct types
df = pd.read_csv(
    url,
    nrows=100,
    dtype=dtype,
    parse_dates=parse_dates
)

# Check types
df.dtypes
```

```python
from sqlalchemy import create_engine
```

```python
# Create database connection
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
#test engine
engine.connect()
```

```python
# See the SQL that pandas will generate
schema_sql = pd.io.sql.get_schema(df, name='yellow_taxi_data', con=engine)
print(schema_sql)
```

```python
# Create table with schema only (no data yet)
df.head(0).to_sql(
    name='yellow_taxi_data',
    con=engine,
    if_exists='replace'
)

print("✅ Table created!")
```

```python
# Create an iterator that reads 100,000 rows at a time
df_iter = pd.read_csv(
    url,
    dtype=dtype,
    parse_dates=parse_dates,
    iterator=True,           # Makes it an iterator!
    chunksize=100000         # Read 100k rows at a time
)

print("✅ Iterator created!")
print("Ready to load data in chunks of 100,000 rows")
```

```python
# Get first chunk
first_chunk = next(df_iter)

print(f"Loaded {len(first_chunk):,} rows")
print(first_chunk.head())
```

```python
from time import time

# Insert first chunk
t_start = time()

first_chunk.to_sql(
    name='yellow_taxi_data',
    con=engine,
    if_exists='append'  # APPEND (don't replace!)
)

t_end = time()

print(f"✅ Inserted {len(first_chunk):,} rows in {t_end - t_start:.2f} seconds")
```

```python
from tqdm.auto import tqdm

# Get total chunks (approximately)
# Full file is ~1.3M rows, we're doing 100k chunks = ~13 chunks
print("Loading remaining chunks...")
print("This will take 5-10 minutes. ☕ Grab some coffee/water!")

chunk_count = 1

for df_chunk in tqdm(df_iter, desc="Loading chunks"):
    t_start = time()
    
    # Insert chunk
    df_chunk.to_sql(
        name='yellow_taxi_data',
        con=engine,
        if_exists='append'
    )
    
    t_end = time()
    chunk_count += 1
    
    print(f"Chunk {chunk_count}: {len(df_chunk):,} rows in {t_end - t_start:.2f} seconds")

print(f"\n🎉 DONE! Loaded {chunk_count} chunks total!")
```

```python
# Count total rows
result = pd.read_sql("SELECT COUNT(*) FROM yellow_taxi_data", engine)
print(f"Total rows: {result.iloc[0, 0]:,}")
```

```python
# Average trip distance
query = "SELECT AVG(trip_distance) as avg_distance FROM yellow_taxi_data"
pd.read_sql(query, engine)
```

```python
# Trips by day
query = """
SELECT 
    DATE(tpep_pickup_datetime) as day,
    COUNT(*) as trips
FROM yellow_taxi_data
GROUP BY day
ORDER BY day
LIMIT 10
"""
pd.read_sql(query, engine)
```
